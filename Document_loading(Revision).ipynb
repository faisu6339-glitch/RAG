# Program 5: Convert raw text into LangChain Document objects.

```python
# Step 1: Import Document
from langchain_core.documents import Document

```

```python
# Step 2: Create a raw text
raw_text = [
    "Python is a programming language.",
    "Machine Learning is a subset of AI.",
    "Deep Learning uses neural networks.",
    "RAG combines retrieval with generation.",
    "Vector databases store embeddings."
]
```

```python
# Step 3: Convert raw text into Document objects
documents = []

for i, text in enumerate(raw_text):
    document = Document(
        page_content=text,
        metadata={"source": f"raw_text_{i}"}
    )
    documents.append(document)
```

```python
# Step 4: Display the documents
for doc in documents:
    print(doc)
```

```python

from langchain_core.documents import Document

raw_text = [
    "Python is a programming language.",
    "Machine Learning is a subset of AI.",
    "Deep Learning uses neural networks.",
    "RAG combines retrieval with generation.",
    "Vector databases store embeddings."
]

documents = []

for index, text in enumerate(raw_text):

    doc = Document(
        page_content=text,
        metadata={
            "document_id": index + 1,
            "source": f"document_{index + 1}.txt"
        }
    )

    documents.append(doc)

for doc in documents:
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("-------------------")
```

# Program 11 :Build a reusable document ingestion function.

```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader
import os
```

**Step 2: Create the reusable function**

```python
def load_documents(folder_path, file_pattern="*.txt"):

    if not os.path.exists(folder_path):
        raise FileNotFoundError("Folder does not exist!")

    loader = DirectoryLoader(
        folder_path,
        glob=file_pattern,
        loader_cls=TextLoader,
        loader_kwargs={"encoding": "utf-8"}
    )

    documents = loader.load()

    return documents
```

```python
import os

from langchain_community.document_loaders import (
    DirectoryLoader,
    TextLoader
)

def load_documents(folder_path, file_pattern="*.txt"):

    try:

        if not os.path.isdir(folder_path):
            raise FileNotFoundError("Invalid folder path!")

        loader = DirectoryLoader(
            folder_path,
            glob=file_pattern,
            loader_cls=TextLoader,
            loader_kwargs={"encoding": "utf-8"},
            show_progress=True,
            silent_errors=False
        )

        documents = loader.load()

        if not documents:
            print("No documents found!")

            return []

        print(f"Successfully loaded {len(documents)} documents.")

        return documents

    except Exception as e:

        print("Error loading documents:", e)

        return []
```
