```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content="Python supports object-oriented programming.",
        metadata={"subject": "Python", "id": 1}
    ),

    Document(
        page_content="Machine learning uses algorithms to learn from data.",
        metadata={"subject": "ML", "id": 2}
    ),

    Document(
        page_content="DBMS is used to manage structured data.",
        metadata={"subject": "DBMS", "id": 3}
    )
]

for doc in documents:
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("-" * 50)
```

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content="Python is easy to learn.",
        metadata={"source": "python.txt", "category": "Programming"}
    ),

    Document(
        page_content="SQL is used to query databases.",
        metadata={"source": "sql.txt", "category": "Database"}
    ),

    Document(
        page_content="CNN is used in image classification.",
        metadata={"source": "cnn.txt", "category": "Deep Learning"}
    )
]

print("Total documents:", len(documents))

for i, doc in enumerate(documents, start=1):

    print(f"\nDocument {i}")
    print("Source:", doc.metadata["source"])
    print("Category:", doc.metadata["category"])
    print("Text:", doc.page_content)
```

# Program 1: TextLoader – Load a TXT file

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("notes.txt")

documents = loader.load()

print("Total documents:", len(documents))

for doc in documents:

    print("Content:")
    print(doc.page_content)

    print("\nMetadata:")
    print(doc.metadata)
```

# Program 2: Load multiple TXT files using DirectoryLoader

```python
import os

os.makedirs("notes", exist_ok=True)

data = {
    "python.txt": "Python is a programming language.",
    "ml.txt": "Machine learning learns patterns from data.",
    "dl.txt": "Deep learning uses neural networks."
}

for filename, content in data.items():

    with open(f"notes/{filename}", "w") as file:
        file.write(content)
```

```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

loader = DirectoryLoader(
    "notes/",
    glob="**/*.txt",
    loader_cls=TextLoader
)

documents = loader.load()

print("Total documents:", len(documents))

for doc in documents:

    print("\nSource:", doc.metadata["source"])
    print("Content:", doc.page_content)
```
