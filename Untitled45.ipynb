```python
from langchain_nvidia_ai_endpoints import ChatNVIDIA

llm = ChatNVIDIA(
    model="nvidia/nemotron-3.5-lightning-30b-a3b",
    temperature=0,
    max_completion_tokens=1024
)

text = """
Python is a programming language.

Subscribe to our channel.

Python supports object-oriented programming.

Follow us on social media.

Python supports multiple programming paradigms.
"""

prompt = f"""
You are a document preprocessing assistant.

Clean the following document.

Instructions:
1. Remove promotional content.
2. Remove irrelevant advertisements.
3. Preserve all useful technical information.
4. Do not invent new information.
5. Do not summarize the document.
6. Preserve the original meaning.

Document:
{text}

Return only the cleaned document.
"""

response = llm.invoke(prompt)

print(response.content)
```
