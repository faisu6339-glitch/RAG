# Program 4: Convert CSV rows into LangChain Documents

```python
import pandas as pd

from langchain_core.documents import Document

df = pd.read_csv("students.csv")

documents = []

for i, row in df.iterrows():
    page_content = f"{row['Name']}, {row['Subject']}, {row['Marks']}"

    doc=Document(
        page_content=page_content,
        metadata={
            "Name": row["Name"],
            "Subject": row["Subject"],
            "Marks": row["Marks"]
        }
    )

    documents.append(doc)
    print(f"Document {i + 1}:")
    print(doc)
    print("-" * 40)
```

# Program 5: Convert CSV rows into LangChain Documents

```python
import pandas as pd

from langchain_core.documents import Document

df = pd.read_csv("students.csv")

documents = []

for index, row in df.iterrows():

    content = f"""
    Student Name: {row['Name']}
    Subject: {row['Subject']}
    Marks: {row['Marks']}
    """

    doc = Document(
        page_content=content,
        metadata={
            "source": "students.csv",
            "row": index,
            "student": row["Name"]
        }
    )

    documents.append(doc)

print("Total documents:", len(documents))

for doc in documents:

    print(doc.page_content)
    print(doc.metadata)
    print("-" * 40)
```

# Program 5: Convert Excel data into LangChain Documents

```python
import pandas as pd

from langchain_core.documents import Document

df = pd.read_excel("employees.xlsx")

documents = []

for index, row in df.iterrows():

    content = (
        f"Employee Name: {row['Name']}\n"
        f"Department: {row['Department']}\n"
        f"Salary: {row['Salary']}"
    )

    doc = Document(
        page_content=content,

        metadata={
            "source": "employees.xlsx",
            "row": index,
            "employee": row["Name"]
        }
    )

    documents.append(doc)

print("Total documents:", len(documents))

for doc in documents:

    print(doc.page_content)
    print(doc.metadata)
    print("-" * 40)
```

# Program 5: Extract tables from Word documents

```python
from docx import Document

doc = Document()

doc.add_heading("Student Records", level=1)

table = doc.add_table(
    rows=1,
    cols=3
)

table.style = "Table Grid"

header = table.rows[0].cells

header[0].text = "Name"
header[1].text = "Subject"
header[2].text = "Marks"

data = [
    ["Faisal", "Python", "90"],
    ["Rahul", "DBMS", "85"],
    ["Amit", "ML", "92"]
]

for row_data in data:

    cells = table.add_row().cells

    for i, value in enumerate(row_data):

        cells[i].text = value

doc.save("Student_Report.docx")
```

```python
from docx import Document

doc = Document("Student_Report.docx")

for table_index, table in enumerate(doc.tables):

    print("Table:", table_index + 1)

    for row in table.rows:

        row_data = []

        for cell in row.cells:

            row_data.append(cell.text)

        print(row_data)
```

# Program 6: Convert Word paragraphs into LangChain Documents

```python
from docx import Document
from langchain_core.documents import Document as LCDocument

word_doc = Document("ML_Notes.docx")

documents = []

for index, paragraph in enumerate(word_doc.paragraphs):

    text = paragraph.text.strip()

    if text:

        doc = LCDocument(
            page_content=text,

            metadata={
                "source": "ML_Notes.docx",
                "paragraph": index + 1
            }
        )

        documents.append(doc)

print("Total documents:", len(documents))

for doc in documents:

    print(doc.page_content)
    print(doc.metadata)
```
